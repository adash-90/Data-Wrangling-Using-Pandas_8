```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

```python
df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha'],
    'Department': ['CSE', 'CSE', 'ECE', 'ECE'],
    'Marks': [85, 92, 78, 88]
})

print("DataFrame 1:")
display(df1)
```

```python
df2 = pd.DataFrame({
    'Student_ID': [105, 106, 107, 108],
    'Name': ['Arjun', 'Neha', 'Rohan', 'Anjali'],
    'Department': ['CSE', 'ECE', 'CSE', 'ECE'],
    'Marks': [90, 76, 95, 82]
})

print("DataFrame 2:")
display(df2)
```

```python
# Create hierarchical index
multi_df = df1.set_index(['Department', 'Student_ID'])

print("DataFrame with Hierarchical Index:")
display(multi_df)
```

```python
print("CSE Students:")
display(multi_df.loc['CSE'])
```

```python
print("Student ID 102:")
display(multi_df.loc[('CSE', 102)])
```

```python
combined_df = pd.concat(
    [df1, df2],
    ignore_index=True
)

print("Combined DataFrame:")
display(combined_df)
```

```python
marks1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Internal': [20, 22, 18, 21]
})

marks2 = pd.DataFrame({
    'External': [65, 70, 60, 67]
})

combined_columns = pd.concat(
    [marks1, marks2],
    axis=1
)

print("Column-wise Combined DataFrame:")
display(combined_columns)
```

```python
student_details = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha'],
    'Department': ['CSE', 'CSE', 'ECE', 'ECE']
})

student_marks = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Marks': [85, 92, 78, 88],
    'Grade': ['A', 'A+', 'B', 'A']
})

merged_df = pd.merge(
    student_details,
    student_marks,
    on='Student_ID',
    how='inner'
)

print("Merged DataFrame:")
display(merged_df)
```

```python
left_merge = pd.merge(
    student_details,
    student_marks,
    on='Student_ID',
    how='left'
)

display(left_merge)
```

```python
outer_merge = pd.merge(
    student_details,
    student_marks,
    on='Student_ID',
    how='outer'
)

display(outer_merge)
```

```python
marks_df = pd.DataFrame({
    'Student': ['Rahul', 'Priya', 'Aman', 'Sneha'],
    'Maths': [85, 92, 78, 88],
    'Python': [90, 95, 80, 86],
    'DBMS': [82, 89, 75, 91]
})

print("Original DataFrame:")
display(marks_df)

long_df = pd.melt(
    marks_df,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Marks'
)

print("Reshaped DataFrame:")
display(long_df)
```

```python
pivot_df = long_df.pivot(
    index='Student',
    columns='Subject',
    values='Marks'
)

print("Pivoted DataFrame:")
display(pivot_df)
```

```python
sales_df = pd.DataFrame({
    'Department': [
        'CSE', 'CSE', 'CSE',
        'ECE', 'ECE', 'ECE'
    ],
    'Semester': [
        1, 1, 2,
        1, 1, 2
    ],
    'Students': [
        60, 65, 70,
        55, 58, 62
    ]
})

print("Sales/Student Data:")
display(sales_df)

pivot_table = pd.pivot_table(
    sales_df,
    values='Students',
    index='Department',
    columns='Semester',
    aggfunc='mean'
)

print("Pivot Table:")
display(pivot_table)
```

```python
stacked = pivot_df.stack()

print("Stacked DataFrame:")
display(stacked)
```

```python
unstacked = stacked.unstack()

print("Unstacked DataFrame:")
display(unstacked)
```
